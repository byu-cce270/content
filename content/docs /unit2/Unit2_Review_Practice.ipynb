## 1 · Indentation: which loop does this line belong to?

Three beams, two point loads each. The program should print **one total deflection per beam**. Right now the totals keep growing from beam to beam. Fix the indentation so each beam starts from zero and prints once.

```python
# @title Exercise 1 - Indentation
E = 29000      # ksi
I = 300        # in^4
beam_lengths = [120, 180, 240]           # in
loads = [(5, 40), (8, 90)]               # (P kips, a in from the support)

total_defl = 0
for L in beam_lengths:
    for P, a in loads:
        total_defl += P * a**2 * (3*L - a) / (6*E*I)
        print(f"L = {L} in: deflection = {total_defl:.4f} in")
```

## 2 · print vs return

`print` shows a value to a person. `return` hands it back to your code. This function should **return both** the area and the weight, so the last line can use them.

```python
# @title Exercise 2 - print vs return
def slab_area_weight(length_ft, width_ft, thick_in):
    area = length_ft * width_ft
    weight = area * thick_in / 12 * 150      # lb, 150 pcf concrete
    print(area, weight)

area, weight = slab_area_weight(20, 12, 6)
print(f"Area {area} ft^2, weight {weight:,.0f} lb")
```

## 3 · A while loop that ends

A truck hauls up to 12 tons per trip. Count the trips needed for 50 tons. This loop never stops. The variable in the condition has to **change inside the loop**.

```python
# @title Exercise 3 - while loop (careful: this one runs forever until fixed)
remaining = 50     # tons
trips = 0
while remaining > 0:
    load = min(12, remaining)
    trips += 1
    print(f"Trip {trips}: {load} tons")
print("Trips needed:", trips)
```

## 8 · Write a file, then read it back

Save each passing beam to `passing.txt`, then read the file back. The file comes out empty because it is never closed. Use `with open(...)`, which closes it for you.

```python
# @title Exercise 8 - files
beams = {"B1": 0.42, "B2": 0.91, "B3": 0.67}     # deflection, in
limit = 0.75

f = open("passing.txt", "w")
for name, d in beams.items():
    if d <= limit:
        f.write(f"{name},{d}\n")

print(open("passing.txt").read())
```
