### Part 1: Define your functions here

Write four functions:
- `cantilever()` and `simply_supported()` — each takes `load_type`, `load`, `L`, `E`, and `I`, and returns the maximum deflection and maximum moment (use IF/ELIF inside to pick the right formula for the load type).
- `inertia()` — takes the base `b` and height `h` and returns the moment of inertia.
- `sigma_max()` — takes the maximum moment `M`, `y` (half the beam height), and `I`, and returns the maximum bending stress.
